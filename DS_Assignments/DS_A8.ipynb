### **Theory/Background:**
**Apache Hive** is an open-source distributed data warehouse software system built on top of Apache Hadoop. It provides data summarization, ad-hoc querying, and analysis of large datasets stored in Hadoop Distributed File System (HDFS) using an SQL-like declarative language called **HiveQL (Hive Query Language)**.

The core architectural components of Apache Hive include:
- **Hive Client / CLI & Beeline:** Provides command-line and JDBC/ODBC connectivity for submitting queries.
- **Hive Driver & Compiler:** Parses HiveQL statements, verifies syntax and catalog metadata, and compiles declarative SQL into directed acyclic graphs (DAGs) of MapReduce, Apache Tez, or Spark execution jobs.
- **Metastore:** A centralized relational database repository (e.g., Derby, MySQL, PostgreSQL) that stores schema metadata, table definitions, column types, partitions, and HDFS physical storage locations.
- **Execution Engine:** Executes compiled query jobs in parallel across Hadoop worker nodes managed by YARN.

Hive follows the **Schema-on-Read** paradigm: raw data is written directly to storage without immediate verification, and table schema, field delimiters, and serializers/deserializers (SerDe) are dynamically projected when the data is read during query execution.
